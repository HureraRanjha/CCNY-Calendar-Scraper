```python
from selenium import webdriver

url = "https://www.ccny.cuny.edu/registrar/fall"

driver = webdriver.Chrome()

driver.get(url)

print("Page title:", driver.title)
```

```python
from bs4 import BeautifulSoup
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

#Wait for the table to appear
WebDriverWait(driver, 20).until(
    EC.presence_of_element_located((By.CSS_SELECTOR, "table tr"))
)

#Parse HTML from Chrome
soup = BeautifulSoup(driver.page_source)

calendar_table = None

for table in soup.find_all("table"):
    table_text = table.get_text(" ", strip=True)
    
    if "DATES" in table_text and "DAYS" in table_text:
        calendar_table = table
        break

if calendar_table is None:
    raise ValueError("Calendar table not found.")

#Parse calendar table
calendar_rows = calendar_table.find_all("tr")

for row in calendar_rows:
    cells = row.find_all(["td", "th"])
    values = [cell.get_text(" ", strip=True) for cell in cells]
    print(values)
```

```python
import pandas as pd
from datetime import datetime

records = []

for row in calendar_rows:
    cells = row.find_all("td")

    if len(cells) == 3:
        values = [cell.get_text(" ", strip=True) for cell in cells]

        if values[0].upper() == "DATES":
            continue

        records.append({
            "date": values[0],
            "dow": values[1],
            "text": values[2]
        })

df = pd.DataFrame(records)

def convert_date(date_str):
    value = date_str.replace("–", "-").replace("—", "-")
    value = value.split("-")[0].strip()

    if "," in value:
        return datetime.strptime(value, "%B %d, %Y").date()

    return datetime.strptime(value + " 2021", "%B %d %Y").date()


df["date"] = df["date"].apply(convert_date)
df = df.set_index("date")

display(df)
print(type(df.index[0]))

driver.quit()
    
```
